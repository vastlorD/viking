```python
def cadastrar_notas():
    """
    Função para cadastrar as notas do aluno.
    Utiliza um laço de repetição para solicitar as notas até que o usuário
    digite 'fim'. As notas são validadas e armazenadas em uma lista.
    """
    notas = []
    print("--- Cadastro de Notas ---")
    print("Digite as notas do aluno. Digite 'fim' para concluir.")

    while True:
        entrada = input(f"Digite a nota {len(notas) + 1}: ")

        if entrada.lower() == 'fim':
            if not notas:
                print("\nNenhuma nota foi inserida. Encerrando o cadastro.")
            break

        try:
            # Tenta converter a entrada para um número float
            nota = float(entrada)
            # Valida se a nota está entre 0 e 10
            if 0 <= nota <= 10:
                notas.append(nota)
            else:
                print("Entrada inválida. Por favor, insira uma nota entre 0 e 10.")
        except ValueError:
            print("Entrada inválida. Por favor, digite um número ou 'fim'.")

    return notas

def calcular_media(notas):
    """
    Calcula a média de uma lista de notas.
    Retorna 0 se a lista estiver vazia para evitar erro de divisão por zero.
    """
    if not notas:
        return 0

    soma_das_notas = sum(notas)
    quantidade_de_notas = len(notas)
    media = soma_das_notas / quantidade_de_notas
    return media

def determinar_situacao(media):
    """
    Determina a situação do aluno com base na média.
    Utiliza uma estrutura condicional (if/else).
    """
    if media >= 7.0:
        return "Aprovado"
    else:
        return "Reprovado"

def exibir_relatorio(notas, media, situacao):
    """
    Exibe o relatório final com todas as informações.
    """
    print("\n--- Relatório Final do Aluno ---")
    print(f"Notas inseridas: {notas}")
    # Formata a média para exibir apenas 2 casas decimais
    print(f"Média final: {media:.2f}")
    print(f"Situação: {situacao}")
    print("---------------------------------")

def main():
    """
    Função principal que orquestra a execução do programa.
    """
    print("Bem-vindo ao Sistema de Gestão de Notas!")

    # 1. Cadastro de Notas
    lista_de_notas = cadastrar_notas()

    # O programa só continua se alguma nota tiver sido inserida
    if lista_de_notas:
        # 2. Cálculo da Média
        media_final = calcular_media(lista_de_notas)

        # 3. Determinação da Situação
        status_do_aluno = determinar_situacao(media_final)

        # 4. Relatório Final
        exibir_relatorio(lista_de_notas, media_final, status_do_aluno)
    else:
        print("Nenhuma nota foi cadastrada para gerar o relatório.")

# Executa a função principal quando o script é iniciado
if __name__ == "__main__":
    main()
```

```python
def adicionar_notas():
    """
    FUNCIONALIDADE 1: Adicionar Notas
    - Pede ao usuário para inserir notas uma a uma.
    - Utiliza um laço 'while' para continuar pedindo até que o usuário digite 'fim'.
    - Armazena as notas válidas (números entre 0 e 10) em uma lista.
    - Retorna a lista de notas.
    """
    notas = []
    print("--- Inserção de Notas ---")
    print("Digite as notas do aluno. Para terminar, digite 'fim'.")

    while True:
        entrada_usuario = input(f"Insira a nota {len(notas) + 1}: ")

        if entrada_usuario.lower() == 'fim':
            break

        try:
            nota = float(entrada_usuario)
            if 0 <= nota <= 10:
                notas.append(nota)
            else:
                print("Erro: A nota deve ser um valor entre 0 e 10.")
        except ValueError:
            print("Erro: Entrada inválida. Por favor, digite um número ou 'fim'.")

    return notas

def calcular_media(lista_de_notas):
    """
    FUNCIONALIDADE 2: Calcular Média
    - Recebe uma lista de notas.
    - Se a lista não estiver vazia, calcula a soma e divide pela quantidade de notas.
    - Retorna a média calculada ou 0 se a lista estiver vazia.
    """
    if not lista_de_notas:
        return 0.0
    return sum(lista_de_notas) / len(lista_de_notas)

def determinar_situacao(media):
    """
    FUNCIONALIDADE 3: Determinar Situação
    - Recebe a média final do aluno.
    - Utiliza uma estrutura condicional 'if/else' para verificar o critério de aprovação.
    - Retorna 'Aprovado' se a média for 7 ou superior, e 'Reprovado' caso contrário.
    """
    if media >= 7.0:
        return "Aprovado"
    else:
        return "Reprovado"

def exibir_relatorio_final(notas, media, situacao):
    """
    FUNCIONALIDADE 4: Exibir Relatório Final
    - Recebe todos os dados processados: a lista de notas, a média e a situação.
    - Imprime na tela um relatório formatado e de fácil leitura.
    """
    print("\n===================================")
    print("      RELATÓRIO FINAL DO ALUNO     ")
    print("===================================")
    print(f"Notas Lançadas: {notas}")
    print(f"Média Final:    {media:.2f}")
    print(f"Situação:       {situacao}")
    print("===================================\n")

# --- Função Principal que Executa o Sistema ---
def executar_sistema():
    """
    Orquestra a execução de todas as funcionalidades na ordem correta.
    """
    print("Iniciando Sistema de Gestão de Notas...")

    # Passo 1: Adicionar as notas
    notas_do_aluno = adicionar_notas()

    if not notas_do_aluno:
        print("\nNenhuma nota foi inserida. O sistema será encerrado.")
        return

    # Passo 2: Calcular a média
    media_do_aluno = calcular_media(notas_do_aluno)

    # Passo 3: Determinar a situação
    situacao_do_aluno = determinar_situacao(media_do_aluno)

    # Passo 4: Exibir o relatório
    exibir_relatorio_final(notas_do_aluno, media_do_aluno, situacao_do_aluno)

# --- Ponto de Entrada do Programa ---
if __name__ == "__main__":
    executar_sistema()
```
